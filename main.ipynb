```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('spotify_data clean.csv')
df['release_year'] = pd.to_datetime(df['album_release_date'], errors='coerce').dt.year
df = df.dropna(subset=['release_year', 'artist_genres'])
df
```

```python
df['genre_list'] = df['artist_genres'].str.split(', ')
df_exploded = df.explode('genre_list')
```

```python
top_n_genres = df_exploded['genre_list'].value_counts().nlargest(10).index.to_list() + ['k-pop']
df_top_genres = df_exploded[df_exploded['genre_list'].isin(top_n_genres)]
```

```python
genre_yearly_counts = df_top_genres.groupby(['release_year', 'genre_list']).size().unstack(fill_value=0)
genre_yearly_counts
```

```python
genre_proportions = genre_yearly_counts.div(genre_yearly_counts.sum(axis=1), axis=0)
genre_proportions
```

```python
plt.figure(figsize=(12, 6))
genre_proportions.plot(kind='area', stacked=True, alpha=0.8, ax=plt.gca())

plt.title('Proportion of Top 6 Genres Over Time')
plt.xlabel('Year')
plt.ylabel('Proportion')
plt.ylim(0, 1)  # Force the y-axis to be exactly 1
plt.legend(title='Genre', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
top10_yearly = (
    df.sort_values(['release_year', 'track_popularity'], ascending=[True, False])
    .groupby('release_year')
    .head(10)
    .groupby('release_year')['track_duration_min']
    .mean()
    .reset_index()
)

plt.figure(figsize=(10, 6))
plt.plot(top10_yearly['release_year'], top10_yearly['track_duration_min'], marker='o', linestyle='-', color='blue')

# Formatting for clarity
plt.title('Average Duration of Top 10 Popular Tracks by Year')
plt.xlabel('Year')
plt.ylabel('Average Duration (minutes)')
plt.grid(True, linestyle='--', alpha=0.7)
```

```python
df = df.dropna(subset=['release_year', 'artist_popularity'])

yearly_artist_pop = df.groupby(['release_year', 'artist_name'])['artist_popularity'].max().reset_index()

top_10_yearly_artists = (
    yearly_artist_pop.sort_values(['release_year', 'artist_popularity'], ascending=[True, False])
    .groupby('release_year')
    .head(10)
)

avg_top_10_pop = top_10_yearly_artists.groupby('release_year')['artist_popularity'].mean().reset_index()

plt.figure(figsize=(10, 6))
plt.plot(avg_top_10_pop['release_year'], avg_top_10_pop['artist_popularity'], marker='o', linestyle='-', color='purple')
plt.title('Average Popularity of Top 10 Artists Over Time')
plt.xlabel('Year')
plt.ylabel('Average Artist Popularity')
plt.grid(True)
plt.show()
```

```python
df = pd.read_csv('spotify_data clean.csv')
df['release_year'] = pd.to_datetime(df['album_release_date'], errors='coerce').dt.year
df = df.dropna(subset=['release_year', 'album_type'])

# Group by release_year and album_type, then count
type_counts = df.groupby(['release_year', 'album_type']).size().unstack(fill_value=0)

# Normalize the counts to get proportions (so the total height is 1)
type_proportions = type_counts.div(type_counts.sum(axis=1), axis=0)

# Plotting as a 100% stacked area chart
plt.figure(figsize=(12, 6))
type_proportions.plot(kind='area', stacked=True, alpha=0.7)

plt.title('Proportion of Album Types Over Time')
plt.xlabel('Year')
plt.ylabel('Proportion')
plt.legend(title='Album Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
sns.boxplot(x='album_type', y='track_popularity', data=df, palette='viridis')

# 4. Add labels and formatting
plt.title('Distribution of Track Popularity by Album Type')
plt.xlabel('Album Type')
plt.ylabel('Track Popularity')
plt.grid(True, axis='y', linestyle='--', alpha=0.7)

# 5. Display the plot
plt.show()
```
